### Un cálculo real: de milímetros a metros cúbicos

Esta es la conversión más usada en hidrología de cuencas.

Un pluviómetro mide **milímetros**, que es una *altura* de lámina de agua. Para saber cuánta agua cayó en realidad, hay que multiplicar esa altura por el **área** sobre la que cayó.

$$V\ (\text{m}^3) = \frac{P\ (\text{mm})}{1000} \times A\ (\text{m}^2)$$

Veámoslo paso a paso, con una variable por paso. Así, si algo sale mal, sabemos exactamente en cuál línea.

```python
precipitacion_mm = 187.4
area_km2 = 85.0

lamina_m  = precipitacion_mm / 1000    # de mm a metros -> estamos convirtiendo de mm a m
area_m2   = area_km2 * 1000000         # de km2 a m2
volumen_m3 = lamina_m * area_m2

print(volumen_m3)
```

```python
#en lo posible para que nuestro programa sea mas legible y de mayor calidad darle formato a los valores
# f-strings: la forma moderna de armar mensajes
# Se pone una f antes de la comilla, y adentro se meten las variables entre llaves {}.

# Con 1 decimal y separador de miles:
print(f"En la cuenca cayeron {volumen_m3:,.1f} m3")

# En millones de metros cúbicos, que es como se reporta:
print(f"Volumen precipitado: {volumen_m3/1_000_000:.2f} millones de m3")

# Varias variables en una sola línea:
print(f"{precipitacion_mm} mm sobre {area_km2} km2 = {volumen_m3/1e6:.2f} Mm3")
```

```python
#una función un codigo que se puede repetir o invocar para hacer un calculo muchas veces
def lluvia_a_volumen(precip_mm, area_km2):
    """Convierte una lámina de lluvia (mm) sobre un área (km2) a volumen (m3)."""
    volumen_m3 = (precip_mm / 1000) * (area_km2 * 1_000_000)
    return volumen_m3


# Ahora la usamos cuantas veces queramos:
print(lluvia_a_volumen(187.4, 85))
print(lluvia_a_volumen(330, 85))
print(lluvia_a_volumen(1628, 12.5))
```

```python
#funcion totalizar lista
def total_lista(p_lista):
  suma = 0 #inicialzar acumulador y asegurar que empieza en 0
  for elemento in p_lista:
    suma = suma + elemento
  return suma
```
